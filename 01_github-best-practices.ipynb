## Branching

### Default Policy

**Never push to main**

Default workflow should be:  
1. Work on a branch
2. Open a PR
3. Merge
4. Delete the branch

### Why Use Branches

* **Safety**: you don't break main
* **Focus**: each branch is about one thing
* **Teamwork**: many people can work on different features without affecting each other

#### When to open a new branch?

Start a new branch for every shippable piece.

### How to Create and Switch

#### Create a new branch

```
git checkout -b feature/login
```

#### Switch between branches

**Switch to main**
```
git checkout main
```

**Switch to feature/login**
```
git checkout feature/login
```

#### Deleting a branch

```
git branch -d feature/login
```

### Branch Naming Conventions

Keep it short, clear, lowercase, with dashes

Examples:
* `feature/login-page`
* `fix/profile-bug`

## Committing

### Default Policy

**Commit often**

Reason:
1. Smaller commits are easier to debug
2. If something breaks, you can roll back to a working commit
3. Helps understand the history of the project

### Writing Good Commit Messages

#### Examples of a good commit message

```
feat(auth): add login form validation
fix(ui): correct navbar link hover state
```

#### Common Convention

```
type(scope): short summary
```

**Types:**
* **feat**: new feature
* **fix**: bug fix
* **docs**: documentation
* **style**: formatting only
* **refactor**: restructuring code without behaviour change
* **test**: adding/fixing tests
* **chore**: maintenance
